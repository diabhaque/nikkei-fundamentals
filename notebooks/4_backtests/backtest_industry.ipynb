```python
import json

import pandas as pd
import plotly.express as px

from tqdm import tqdm
from collections import defaultdict

from utils.backtest import run_daily_backtest
from utils.data import jp_500
from utils.datapath import industry_elo_signals_path, daily_prices_path, docs_metadata_path
from utils.datetime import get_rebalance_date, get_quarter_end_date_string
```

```python
with open(industry_elo_signals_path) as f:
    industry_elo_signal_series = json.load(f)
    industry_elo_signal_series = {
        industry: { get_rebalance_date(q_string): signal for q_string, signal in signal_series.items() }
        for industry, signal_series in industry_elo_signal_series.items()
    }
```

```python
# Combined portfolio backtest
# TODO: understand why shorting fails

combined_signal_series = defaultdict(dict)
for industry, signal_series in industry_elo_signal_series.items():
    for date, signals in signal_series.items():
        combined_signal_series[date].update(signals)

industry_elo_signal_series["Combined"] = combined_signal_series
```

```python
black_list = ["2372", "9749"]
price_df = pd.read_csv(daily_prices_path)
price_df = price_df.drop(columns=["Unnamed: 0"] + black_list).set_index("Date")

price_df = price_df.ffill()
```

```python
# Individual industry backtests
industry = "Combined"
signal_series = industry_elo_signal_series[industry]
```

```python
timeline = run_daily_backtest(
    price_df, signal_series, shorting_enabled=True
)

navs = timeline.get_navs()
print(f"Final NAV: {navs.iloc[-1]:.2f}")
print(f"Annualized Return: {timeline.calculate_annualized_return():.2%}")
print(f"Sharpe Ratio: {timeline.calculate_sharpe_ratio():.2f}")
```

```python
# container for all timelines
timelines = {}
rebalance_days = 21
```

```python
for industry, signal_series in tqdm(industry_elo_signal_series.items()):
    timelines[industry] = run_daily_backtest(
        price_df,
        signal_series,
        shorting_enabled=True,
    )
```

```python
from utils.dummy import generate_signals

rerun = False
asset_count = 20
num_noise = 100

constituents = {
    get_quarter_end_date_string(quarter): basket for quarter, basket in jp_500.items()
}

# run N noise backtests
if rerun:
    for i in tqdm(range(num_noise)):
        label = f"noise_{i}"
        timelines[label] = run_daily_backtest(
            price_df,
            generate_signals(constituents),
            asset_count=asset_count,
            shorting_enabled=True,
        )
```

```python
industry_cr = [
    pd.Series(timelines[industry].get_cumulative_returns(), name=f"{industry}")
    for industry in industry_elo_signal_series.keys()
]


# noise series
noise_cr = [
    pd.Series(timelines[f"noise_{i}"].get_cumulative_returns(), name=f"noise_{i}")
    for i in range(num_noise)
]

df = pd.concat(industry_cr + noise_cr, axis=1)
```

```python
fig = px.line(
    df,
    y=[col for col in df.columns if col.startswith("noise_")],
    labels={"value": "Cumulative Return", "index": "Date"},
)

# style the noise traces
for trace in fig.data:
    trace.update(line=dict(color="black", width=2, dash="dot"), opacity=0.1)

# overlay the factor line on top
for industry in industry_elo_signal_series.keys():
    fig.add_scatter(
        x=df.index,
        y=df[industry],
        mode="lines",
        name=industry,
        line=dict(width=2),
        marker=dict(size=4),
        visible="legendonly"
    )

fig.update_layout(
    width=1000,
    height=600,
    hovermode="closest",
    showlegend=True,
)

fig.show()
```
