```python
#!/usr/bin/env python3
# compare_earnings.py

import os
import pdfplumber
import openai
import tiktoken
from typing import List

# Set your OpenAI API key in your environment:
#   export OPENAI_API_KEY="your_api_key_here"
openai.api_key = os.getenv("OPENAI_API_KEY")

def extract_text(path: str) -> str:
    """
    Extract all text from a PDF file.
    """
    text_parts: List[str] = []
    with pdfplumber.open(path) as pdf:
        for page in pdf.pages:
            page_text = page.extract_text()
            if page_text:
                text_parts.append(page_text)
    return "\n".join(text_parts)

def chunk_text(text: str, max_tokens: int = 2000) -> List[str]:
    """
    Split text into chunks not exceeding max_tokens (approximate).
    """
    encoder = tiktoken.encoding_for_model("gpt-4o-mini")
    tokens = encoder.encode(text)
    chunks: List[str] = []
    for i in range(0, len(tokens), max_tokens):
        chunk = tokens[i : i + max_tokens]
        chunks.append(encoder.decode(chunk))
    return chunks

def summarize_pdf(path: str, label: str) -> str:
    """
    Extracts and summarises key financial metrics from a PDF:
    revenue, operating profit, EBITDA, cash flow, etc.
    """
    raw_text = extract_text(path)
    chunks = chunk_text(raw_text)
    summaries: List[str] = []

    for idx, chunk in enumerate(chunks, start=1):
        prompt = (
            f"{label} の第X四半期決算報告書セクション {idx} を要約し、"
            "重要な財務指標（売上高、営業利益、EBITDA、キャッシュフローなど）を"
            "箇条書きで抽出してください：\n\n"
            + chunk
        )
        resp = openai.ChatCompletion.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": "あなたは日本企業の財務分析の専門家です。"},
                {"role": "user",   "content": prompt}
            ],
            temperature=0.0,
            max_tokens=500
        )
        summaries.append(resp.choices[0].message.content.strip())

    return "\n\n".join(summaries)

def compare_companies(path_a: str, path_b: str) -> str:
    """
    Summarises two PDFs and asks the model which company is more likely to succeed.
    """
    summary_a = summarize_pdf(path_a, "企業A")
    summary_b = summarize_pdf(path_b, "企業B")

    compare_prompt = (
        "以下は2社の四半期決算報告書の要約です。どちらの企業が今後より成功する可能性が高いか、"
        "売上成長率、利益率、キャッシュポジション、その他重要指標を踏まえて日本語で比較・結論を述べてください。\n\n"
        "【企業Aの要約】\n" + summary_a + "\n\n"
        "【企業Bの要約】\n" + summary_b + "\n\n"
        "結論："
    )

    result = openai.ChatCompletion.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "あなたは金融アナリストです。"},
            {"role": "user",   "content": compare_prompt}
        ],
        temperature=0.0,
        max_tokens=600
    )
    return result.choices[0].message.content.strip()

if __name__ == "__main__":
    import argparse

    parser = argparse.ArgumentParser(
        description="Compare two Japanese quarterly earnings PDFs and determine which company is likely to succeed."
    )
    parser.add_argument("pdf_a", help="Path to the first PDF (企業A).")
    parser.add_argument("pdf_b", help="Path to the second PDF (企業B).")
    args = parser.parse_args()

    comparison = compare_companies(args.pdf_a, args.pdf_b)
    print("\n--- 比較結果 ---\n")
    print(comparison)
```
